# Exploração inicial: avaliações Olist
Este notebook descreve distribuições básicas de notas e textos. Não executa treinamento automaticamente.
Fonte: https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce (CC BY-NC-SA 4.0).

```python
from pathlib import Path
import pandas as pd
path = Path('../data/raw/olist_order_reviews_dataset.csv')
if not path.exists():
    raise FileNotFoundError(f'Baixe o CSV da Olist em: {path}')
df = pd.read_csv(path)
print('Linhas e colunas:', df.shape)
df[['review_score', 'review_comment_message']].head()
```

```python
print('Distribuição das estrelas:')
print(df['review_score'].value_counts(dropna=False).sort_index())
print('Comentários ausentes:', int(df['review_comment_message'].isna().sum()))
print('Comprimento (caracteres):')
print(df['review_comment_message'].dropna().astype(str).str.len().describe())
```

## Interpretação e limitações
A distribuição de estrelas pode estar desbalanceada. Estrelas são proxies de sentimento; a existência de comentários ambíguos exige análise adicional. O modelo principal é treinado separadamente pelo comando `python -m src.train` a partir da raiz do projeto.
